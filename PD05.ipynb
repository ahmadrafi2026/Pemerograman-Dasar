```python
#Nama : Ahmad Rafi'i Malikul Rauf
#NIM  : 4232501036

jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

total = jumlah_buku * harga_buku

diskon = total * diskon_persen // 100

bayar = total - diskon

kembalian = uang_budi - bayar

per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

uang_cukup = uang_budi >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
#Nama : Ahmad Rafi'i Malikul Rauf
#NIM  : 4232501036

jumlah_panel = 12
daya_panel = 450
penyinaran = 5
efisiensi = 0.8
kebutuhan = 25
tarif = 1444.70

energi_hari = jumlah_panel * daya_panel * penyinaran * efisiensi / 1000

energi_bulan = energi_hari * 30
nilai_rupiah = energi_bulan * tarif

kekurangan = kebutuhan - energi_hari

cukup = energi_hari >= kebutuhan

print("Energi per hari       :", energi_hari, "kWh")
print("Energi per bulan      :", energi_bulan, "kWh")
print("Nilai listrik         : Rp", round(nilai_rupiah, 2))
print("Kekurangan energi     :", kekurangan, "kWh/hari")
print("PLTS mencukupi?       :", cukup)
```

```python
#Nama : Ahmad Rafi'i Malikul Rauf
#NIM  : 4232501036

daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

energi = daya_generator * jam_operasi

solar_harian = energi * konsumsi_solar
biaya_solar = solar_harian * harga_solar

hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

rugi_daya = arus ** 2 * hambatan

rugi_tinggi = rugi_daya > 150 and arus >= 20

print("Energi per hari       :", energi, "kWh")
print("Solar per hari        :", solar_harian, "liter")
print("Biaya solar per hari  : Rp", biaya_solar)
print("Tangki cukup          :", hari_penuh, "hari penuh")
print("Sisa solar            :", sisa_solar, "liter")
print("Rugi daya kabel       :", rugi_daya, "W")
print("Rugi kabel tinggi?    :", rugi_tinggi)
```
