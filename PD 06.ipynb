```python

n = int(input("Masukkan bilangan bulat (1-5): "))

if n < 1 or n > 5:
    print("Input tidak valid! Masukkan angka 1 sampai 5.")
elif n == 1:
    faktorial = 1
    print("Faktorial dari", n, "adalah", faktorial)
elif n == 2:
    faktorial = 2 * 1
    print("Faktorial dari", n, "adalah", faktorial)
elif n == 3:
    faktorial = 3 * 2 * 1
    print("Faktorial dari", n, "adalah", faktorial)
elif n == 4:
    faktorial = 4 * 3 * 2 * 1
    print("Faktorial dari", n, "adalah", faktorial)
else:
    faktorial = 5 * 4 * 3 * 2 * 1
    print("Faktorial dari", n, "adalah", faktorial)
```

```python

faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

skor = faktor_daya * efisiensi

print("Skor kualitas daya:", skor)

if skor >= 85 and faktor_daya >= 0.90:
    print("Klasifikasi: PRIMA - sistem dalam kondisi optimal")
elif skor >= 70 and faktor_daya >= 0.80:
    print("Klasifikasi: BAIK - sistem memadai, pantau rutin")
elif skor >= 50 or faktor_daya >= 0.75:
    print("Klasifikasi: CUKUP - ada ruang perbaikan")
else:
    print("Klasifikasi: BURUK - perlu perbaikan segera")
```

```python

iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc = float(input("Masukkan SoC baterai (%): "))

daya_panel = iradiasi * 0.18 * 10

print("Estimasi daya panel:", daya_panel, "W")

if iradiasi > 600 and soc < 80:
    print("Mode: CHARGING - prioritaskan pengisian baterai")
elif iradiasi > 600 and soc >= 80:
    print("Mode: GRID-EXPORT - ekspor kelebihan daya ke jaringan")
elif iradiasi > 200 and soc >= 20:
    print("Mode: NORMAL - operasi mandiri")
elif iradiasi <= 200 and soc >= 50:
    print("Mode: BATTERY-ONLY - andalkan baterai")
else:
    print("Mode: EMERGENCY - kurangi beban")
```
