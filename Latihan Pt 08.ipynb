```python
#6H

energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for i in range(len(energy)):
    if run_hours[i] < 24:
        print("Day", days[i], ":", run_hours[i], "h,", energy[i], "MWh")

weekday = []
weekend = []

for i in range(len(energy)):
    if DAY_NAMES[i % 7] == "Sat" or DAY_NAMES[i % 7] == "Sun":
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])

print("Weekday average:", round(sum(weekday) / len(weekday)), "MWh")
print("Weekend average:", round(sum(weekend) / len(weekend)), "MWh")
```

```python
#6i

temps = [28, 31, 30, 33, 29, 27, 32]

print("3 hari pertama:", temps[:3])
print("2 hari terakhir:", temps[-2:])

temps.append(34)
print("Setelah append:", temps)

temps.pop()
print("Setelah pop:", temps)

print("Urutan suhu:", sorted(temps))
print("Data asli:", temps)

average = sum(temps) / len(temps)
highest = max(temps)
day = temps.index(highest) + 1

print("Rata-rata:", average)
print("Suhu tertinggi:", highest, "°C")
print("Hari ke-", day)

above_average = [temp for temp in temps if temp > average]
print("Suhu di atas rata-rata:", above_average)
```

```python
#Latihan 1

scores = [80, 65, 90, 75, 70, 85]

print("Nilai pertama:", scores[0])
print("Nilai terakhir:", scores[-1])

average = sum(scores) / len(scores)
print("Rata-rata:", average)

count = 0
for score in scores:
    if score > average:
        count += 1

print("Nilai di atas rata-rata:", count)

lowest = min(scores)
student_number = scores.index(lowest) + 1
print("Nomor siswa dengan nilai terendah:", student_number)

new_scores = [score + 5 for score in scores]
print("Nilai setelah ditambah 5:", new_scores)
```

```python
#Latihan 2

energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

days = list(range(1, 31))

total = sum(energy)
average = total / len(energy)

print("TOTAL ENERGI :", total, "MWh")
print("RATA-RATA    :", round(average, 2), "MWh")

highest = sorted(
    enumerate(energy, start=1),
    key=lambda x: x[1],
    reverse=True
)

lowest = sorted(
    enumerate(energy, start=1),
    key=lambda x: x[1]
)

print("\n3 HARI TERTINGGI:")
for day, value in highest[:3]:
    print("Hari", day, ":", value, "MWh")

print("\n3 HARI TERENDAH:")
for day, value in lowest[:3]:
    print("Hari", day, ":", value, "MWh")

target = float(input("\nMasukkan target energi (MWh): "))

below_target = []
for i in range(len(energy)):
    if energy[i] < target:
        below_target.append(days[i])

print("Jumlah hari di bawah target:", len(below_target))
print("Hari di bawah target:", below_target)

limit = average * 0.97

print("\n7-DAY MOVING AVERAGE DI BAWAH 97%:")
for i in range(6, len(energy)):
    window = energy[i-6:i+1]
    avg_7 = sum(window) / 7

    if avg_7 < limit:
        print("Hari", i + 1, ":", round(avg_7, 2), "MWh")

print("\nTOTAL ENERGI PER MINGGU:")
for i in range(4):
    start = i * 7
    end = start + 7
    weekly_total = sum(energy[start:end])
    print("Minggu", i + 1, ":", weekly_total, "MWh")

remaining = sum(energy[28:30])
print("Sisa hari (29-30):", remaining, "MWh")
```
