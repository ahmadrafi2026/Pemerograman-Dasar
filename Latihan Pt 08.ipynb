```python
#6F

scores = [80, 65, 90, 75, 70]

total = 0
count_pass = 0

for score in scores:
    total = total + score

    if score >= 75:
        count_pass = count_pass + 1

    print("running total:", total)

average = total / len(scores)

print("Total   :", total)
print("Average :", average)
print("Scores >= 75:", count_pass, "of", len(scores))
```

```python
#6G

temps = [28, 31, 30, 33, 29, 27, 32]

avg5 = []
for i in range(4, len(temps)):
    window = temps[i - 4 : i + 1]
    avg5.append(round(sum(window) / 5, 1))

print("5-day averages:", avg5)
print(len(temps), "values ->", len(avg5), "averages")
```

```python
#6H

energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

days = list(range(1, 31))
run_hours = [24] * 30
run_hours[16] = 18
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for i in range(len(energy)):
    if run_hours[i] < 24:
        print("Day", days[i], ":", run_hours[i], "h,", energy[i], "MWh")

weekday = []
weekend = []

for i in range(len(energy)):
    if DAY_NAMES[i % 7] == "Sat" or DAY_NAMES[i % 7] == "Sun":
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])

print("Weekday average:", round(sum(weekday) / len(weekday)), "MWh")
print("Weekend average:", round(sum(weekend) / len(weekend)), "MWh")
```

```python
#6i

temps = [28, 31, 30, 33, 29, 27, 32]

print("3 hari pertama:", temps[:3])
print("2 hari terakhir:", temps[-2:])

temps.append(34)
print("Setelah append:", temps)

temps.pop()
print("Setelah pop:", temps)

print("Urutan suhu:", sorted(temps))
print("Data asli:", temps)

average = sum(temps) / len(temps)
highest = max(temps)
day = temps.index(highest) + 1

print("Rata-rata:", average)
print("Suhu tertinggi:", highest, "°C")
print("Hari ke-", day)

above_average = [temp for temp in temps if temp > average]
print("Suhu di atas rata-rata:", above_average)
```
