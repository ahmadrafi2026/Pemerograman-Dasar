```python
#Soal 7

suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    #Cek apakah suhu mencapai tepat 80 °C
    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

Program ini bekerja dengan menetapkan suhu awal sebesar 60 °C, lalu menjalankan perulangan *while* yang terus menaikkan suhu sebesar 5 derajat dan mencetak nilai terbarunya selama suhu masih di bawah 85 °C. Di dalam perulangan tersebut, terdapat kondisi *if* yang secara otomatis mencetak peringatan WARNING: Mendekati target saat suhu mencapai tepat 80 °C. Ketika suhu menyentuh angka 85 °C, kondisi perulangan menjadi tidak terpenuhi sehingga program keluar dari loop dan diakhiri dengan mencetak pesan "Target 85 C tercapai.

```python
#Soal 9

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for i, nilai_str in enumerate(data, start=1):
    try:
        #Ubah data menjadi float dengan try-except ValueError
        nilai = float(nilai_str)
    except ValueError:
        print(f"Unit {i}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    #Jika nilai melebihi 2.0
    if nilai > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")

    #Jika nilai melebihi 3.0
    if nilai > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

Program ini memeriksa data vibrasi mesin menggunakan perulangan *for*, di mana konversi teks ke angka dilindungi oleh *try-except* untuk menangani data rusak ERR dengan mencatatnya sebagai data tidak valid dan melewatinya. Jika data valid dan nilainya melebihi 2.0, program memunculkan peringatan vibrasi tinggi, serta langsung menghentikan pemeriksaan secara total menggunakan *break* jika nilai melebihi 3.0. Di akhir, program mencetak total keseluruhan data yang tidak valid.

```python
#Challlenge

data_mentah = ["12.5", "15.0", "rusak", "18.2", "9.4"]

print("=== LAPORAN KONSUMSI LISTRIK ===")

# Perulangan for & enumerate untuk menelusuri data dengan nomor hari
for hari, nilai_str in enumerate(data_mentah, start=1):
    try:
        #mengubah teks menjadi angka float
        kwh = float(nilai_str)

        # Percabangan if-elif-else untuk status penggunaan
        if kwh > 20.0:
            status = "Boros!"
        elif kwh > 15.0:
            status = "Cukup tinggi"
        else:
            status = "Normal"

        print(f"Hari ke-{hari}: {kwh} kWh - Status: {status}")

    except ValueError:
        print(f"Hari ke-{hari}: Data '{nilai_str}' tidak valid (diabaikan)")
```


Penjelasan Komponen Program:

1.List data_mentah: Menyimpan kumpulan data konsumsi listrik dalam bentuk teks string, di mana terdapat data kotor/rusak rusak untuk menguji ketahanan program.
2.Perulangan for & enumerate: Berfungsi menelusuri setiap data di dalam list satu per satu sekaligus memberikan nomor urut hari secara otomatis mulai dari angka 1.
3.try-except ValueError: Berperan sebagai pengaman error handling. Jika program mendapati data yang tidak bisa diubah menjadi angka, program tidak akan crash (berhenti mendadak), melainkan langsung mencetak pesan peringatan dan melanjutkan ke hari berikutnya.
4.Percabangan if-elif-else: Menentukan status penggunaan listrik (Boros, Cukup tinggi, atau Normal) hanya pada data angka yang berhasil lolos validasi try-except.
